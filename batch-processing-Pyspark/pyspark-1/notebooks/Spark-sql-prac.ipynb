```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Day4") \
    .getOrCreate()
```

```python
df = spark.read.csv("employee_data.csv", header=True, inferSchema=True)
df.show()
```

```python
df.explain()
```

```python
df.createOrReplaceTempView("employees_view")
```

```python
spark.sql("SELECT * FROM employees_view LIMIT 5").show()
```

```python
spark.sql("SELECT name, age FROM employees_view WHERE age > 30").show()
```

```python
spark.sql("SELECT department, AVG(salary) AS avg_salary FROM employees_view GROUP BY department").show()
```

```python
spark.sql("SELECT * FROM employees_view WHERE age > 30").show()
```

```python
from pyspark.sql.functions import col

df.filter(col("age") > 30).show()
```

```python
spark.sql("SELECT department, AVG(salary) AS avg_salary FROM employees_view GROUP BY department").show()
```

```python
from pyspark.sql.functions import avg

df.groupBy("department").agg(avg("salary").alias("avg_salary")).show()
```

```python
spark.sql("""
SELECT department, AVG(salary) AS avg_salary
FROM employees_view
WHERE age > 25
GROUP BY department
HAVING AVG(salary) > 6000
""").show()
```

```python
from pyspark.sql.functions import avg

df_filtered = df.filter(col("age") > 25)
df_grouped = df_filtered.groupBy("department").agg(avg("salary").alias("avg_salary"))
df_grouped.filter(col("avg_salary") > 6000).show()
```

```python
spark.sql("""
SELECT a.name AS name_a, b.name AS name_b
FROM employees_view a
INNER JOIN employees_view b
ON a.age = b.age AND a.name != b.name
""").show()
```

```python
df_a = df.alias("a")
df_b = df.alias("b")
df_j= df_a.join(df_b, (df_a.age == df_b.age) & (df_a.name != df_b.name), "inner")
df_j.show()
```

```python
df_j.explain()
```

```python
spark.sql("""
SELECT name, salary,
       RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS salary_rank
FROM employees_view
""").show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import rank, desc

window_spec = Window.partitionBy("department").orderBy(desc("salary"))
df.withColumn("salary_rank", rank().over(window_spec)).show()
```
